## Introduction

This notebook explores the following datasets gathered from the Kaggle website.

- ``short_video_engagement.csv``
- ``tiktok_dataset.csv``
- ``youtube_data.csv``

We will be using these datasets to look into how video length affects different forms of engagement on social media platforms including Instagram, YouTube, and TikTok. Our motivation for this project comes from our interest in how short-form content has dominated the internet.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
shortvids = pd.read_csv('data/short_video_engagement.csv')
print('Shape:', shortvids.shape)
shortvids.head()
```

```python
shortvids.describe()
```

```python
shortvids.columns
```

```python
shortvids.dtypes
```

Since we're not interested in ``edge_intensity``, ``color_histogram``, ``spectral_entropy``, ``audio_intensity`` we will drop these columns.

```python
shortvids = shortvids.drop(columns = ['edge_intensity', 'color_histogram', 'spectral_entropy', 'audio_intensity'])
```

```python
# confirm that the columns are dropped
print(shortvids.shape)
shortvids.head()
```

```python
# check for nulls
shortvids.isna().sum().sum()
```

Luckily, there are no null values in any columns, meaning that our data is clean and ready for visualization.
