```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
```

```python
tiktok = pd.read_csv("tiktok_dataset.csv")
```

```python
tiktok.columns
```

```python
tiktok_std = tiktok.rename(columns={
    "video_view_count": "views",
    "video_like_count": "likes",
    "video_duration_sec": "duration"
})[["video_id", "views", "likes", "duration"]]
```

```python
tiktok_clean = tiktok_std.dropna()
tiktok_clean
```

```python
tiktok_clean["engagement_rate"] = tiktok_clean["likes"] / tiktok_clean["views"]
tiktok_clean["engagement_rate"] = tiktok_clean["engagement_rate"] * 100
# standardizing
tiktok_clean["log_views"] = np.log1p(tiktok_clean["views"])
tiktok_clean["log_likes"] = np.log1p(tiktok_clean["likes"])
```

```python
tiktok_clean["duration_category"] = pd.qcut(
    tiktok_clean["duration"],
    q=3,
    labels=["short", "medium", "long"]
)
```

```python
tiktok_clean.groupby("duration_category")["duration"].max()
```
